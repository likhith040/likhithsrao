```python
 #creating a DataFrame
data_dict={
    'Name': ['Alice', 'Bob', 'Charlie', 'David','Kartik','Kartik','Harshali','Yuvraj'],
    'Age': [24, 25, 22,4,17,15,20,20],
    'City': ['New York', 'New Jersey', 'London', 'Tokyo','Mumbai','Chicago','Washington','Washington']
}
df=pd.DataFrame(data_dict)
print(df)
print(df.head())
print(df.info())
print(df.describe())
```

```python
print("single column:\n",df['City'])
print("multiple column:\n",df['Age','Name'])
```
