```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
tips=sns.load_dataset('tips')
print(tips)
plt.figure(figsize=(6,4))
sns.scatterplot(x='total_bill',y='tip',data=tips,hue="sex",style="time")
plt.title("Scatter Plot of Total Bill vs Tip")
plt.xlabel("Total Bill")
plt.ylabel("Tip")
plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sms
tips=sms.load_dataset('tips')
print(tips)
plt.figure(figsize=(6,4))
sms.lineplot(y='total_bill',x='size',data=tips,ci=None)
plt.title("Line Plot of Total Bill vs Tip")
plt.ylabel("Total Bill")
plt.xlabel("size")
plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sms
tips=sms.load_dataset('tips')
print(tips)
plt.figure(figsize=(6,4))
sms.barplot(y='total_bill',x='day',data=tips,hue='sex')
plt.title("Bar Plot of Total Bill vs Tip")
plt.ylabel("Total Bill")
plt.xlabel("day")
plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
tips=sms.load_dataset('tips')
print(tips)
plt.figure(figsize=(6,4))
sns.countplot(x='day',data=tips,hue='sex')
plt.title("Count Plot of Total Bill vs Tip")
plt.ylabel("Total Bill")
plt.xlabel("day")
plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sms
tips=sms.load_dataset('tips')
plt.figure(figsize=(6,4))
sms.boxplot(y='total_bill',x='day',data=tips,hue='sex')
plt.title("Box Plot of Total Bill vs Tip")
plt.ylabel("Total Bill")
plt.xlabel("day")
plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sms
tips=sms.load_dataset('tips')
plt.figure(figsize=(6,4))
sms.violinplot(y='total_bill',x='day',data=tips,hue='sex',split=True)
plt.title("Violin Plot of Total Bill vs Tip")
plt.ylabel("Total Bill")
plt.xlabel("day")
plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sms
tips=sms.load_dataset('tips')
plt.figure(figsize=(6,4))
corr=tips.corr(numeric_only=True)
sms.heatmap(corr,annot=True,cmap='cool',linewidths=0.5)
plt.title("Heatmap of Total Bill vs Tip")
plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sms
tips=sms.load_dataset('tips')
plt.figure(figsize=(6,4))
sms.pairplot(tips,hue='sex')
plt.title("Pair Plot of Total Bill vs Tip")
plt.show()
```
